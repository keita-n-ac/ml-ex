1. 乱数を利用して，プログラムで10000回サイコロを振ったときに，各目がそれぞれ何回出たのかを出力するプログラムを作成せよ

```python
import random
count = [0, 0, 0, 0, 0, 0] # count[i]がi-1の目をカント担当

for i in range(10000):
    dice =  random.randint(1, 6)
    count[dice-1] += 1

for i in range(len(count)):
    print(i+1, 'の目が出た回数:', count[i], '回')
```

2. サンプルプログラムを参考にして，比例のグラフ，反比例のグラフ，1次関数のグラフ，2次関数のグラフを作成せよ．ただし（0 < x <= 10とすること）

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0.01, 10.01, 0.01)
y1 = 2 * x # 比例
y2 = 2 / x # 反比例
y3 = 2 * x + 100 # 1次関数
y4 = 2 * x * x # 2次関数

plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)
plt.plot(x, y4)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
