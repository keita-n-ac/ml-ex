### 演習問題
1. 以下の表に対して，横軸を区名，縦軸を人口数とした積み立て棒グラフを作成せよ

| Area       | Male   | Female |
|------------|--------|--------|
| Kita-ku    | 136596 | 152727 |
| Minami-ku  | 62347  | 73430  |
| Nishi-ku   | 100027 | 117013 |
| Higashi-ku | 126023 | 139356 |

```python
import matplotlib.pyplot as plt

num = [1, 2, 3, 4]
male = [136596, 62347, 100027, 126023]
female = [152727, 73430, 117013, 139356]
area = ['Kita-ku', 'Minami-ku', 'Nishi-ku', 'Higashi-ku']

plt.bar(num, male, tick_label=area, color='b')
plt.bar(num, female, bottom=male, color='r')
plt.legend(['Male', 'Female'])
plt.show()
```

2. 以下の指示に従って，プログラムを作成せよ

- Step1: サンプルプログラムを利用して，-πからπの範囲を要素数が100となるように分割したデータdataを作成する（π: 円周率）
- Step2: dataの要素値を余弦(cos)値に変換したものをデータxとする
- Step3: dataの要素値を正弦(sin)値に変換したものをデータyとする
- Step4: データxとデータyを散布図でプロットを行うと円状にプロットされる

```python
import numpy as np
import matplotlib.pyplot as plt
data = np.linspace(-np.pi, np.pi, 100)
x = np.sin(data)
y = np.cos(data)
plt.scatter(x, y, marker='*', color='b')
plt.show()
```

3. 以下の指示に従って，プログラムを作成せよ
- Step1: データ1（data1）を np.random.randn(1000)，データ2（data2）を np.random.randn(1000)で作成しなさい
- Step2: 横軸をデータ1の値，縦軸をデータ2の値として散布図を作成する（ただし，データ1の値，データ2の値が両方0以上の場合は，マーカーと色を変えて表示すること）
  - ヒント: データ1とデータ2を，両方0以上の場合と，そうでない場合にデータを分割する

```python
import numpy as np
import matplotlib.pyplot as plt
data1 = np.random.randn(1000)
data2 = np.random.randn(1000)

x1 = []
x2 = []
y1 = []
y2 = []
for i in range(1000):
    if data1[i] >= 0 and data2[i] >= 0:
        x1.append(data1[i])
        y1.append(data2[i])
    else:
        x2.append(data1[i])
        y2.append(data2[i])

plt.scatter(x1, y1, marker='*', color='r')
plt.scatter(x2, y2, marker='D', color='b')
plt.show()
```
